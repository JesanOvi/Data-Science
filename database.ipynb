```python
#Pull in the libraries we need
import pandas as pd
from pandas import DataFrame

#Adding in a few new database libraries for connection and inspection
import sqlite3
from sqlalchemy import create_engine, inspect
```

```python
dburi = 'sqlite:///dsci503.sqlite3'
```

```python
#We setup an inspector object and engine to see what is in our file
#A sqlite3 or database file cannot be viewed like a text or .csv can
inspector = inspect(create_engine(dburi))

#Returns a list of databases and tables that are included in the file
inspector.get_schema_names()
inspector.get_table_names('main')
```
