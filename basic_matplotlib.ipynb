```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3]
y = [10, 4, 50]
```

```python

plt.plot(x, y)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("My Graph")

plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [120, 135, 150, 142, 170]

plt.plot(months, sales)

plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Monthly Sales")

plt.show()
```

```python
hours = [1, 2, 3, 4, 5, 6, 7, 8]
scores = [55, 60, 65, 68, 70, 72, 85, 90]

plt.scatter(hours, scores)

plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs. Exam Score")

plt.show()
```

```python
ages = [
    18, 19, 19, 20, 20, 21, 21, 21,
    22, 22, 23, 23, 24, 25, 25, 26,
    27, 28, 29, 30
]

plt.hist(ages, bins=5)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")

plt.show()
```

```python
import random

age = random.sample(range(20, 70), 10)

print(age)
```

```python
income = random.sample(range(5000, 25000), 10)

print(income)
```

```python
plt.scatter(age, income)

plt.xlabel("age")
plt.ylabel("income")
plt.title("age vs. income")

plt.show()
```
