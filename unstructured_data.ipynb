Kaggle Dataset

```python
tm_df = pd.read_csv("IOT-temp.csv")
```

```python
tm_df
```

```python
tm_df.columns
```

```python
plt.figure(figsize=(15,5))
plt.plot(tm_df['temp'])
plt.title('Temperature Over Time')
plt.xlabel('Time')
plt.ylabel('Temperature (°C)')
plt.show()
```

```python
temIn = tm_df[tm_df['out/in'] == 'In']
temOut = tm_df[tm_df['out/in'] == 'Out']
```

```python
temOut['temp']
```

```python
plt.figure(figsize=(15,5))
plt.plot(temOut['temp'], color='red')
plt.plot(temIn['temp'], color='blue')
plt.title('Temperature Over Time')
plt.xlabel('Time')
plt.ylabel('Temperature (°C)')
plt.legend(['Outside', 'Inside'])
plt.show()
```
